```python
from vpython import *
```

```python
scene = canvas()

scene.autoscale=0 

aU = 1.5e11 
g = 6.67e-11
mSun = 2e30
mEarth = 6e24
mJupiter = 1.89e27
yR = 3.15e7 #Length of the year in seconds 

scene.range=6*aU


earth = sphere(pos = vector(aU,0,0), radius = .1*aU, texture=textures.earth,)
sun = sphere(pos = vector(0,0,0), radius = .5*aU, color=color.yellow)
jupiter = sphere(pos = vector(5*aU,0,0), radius = .2*aU, color=color.orange)

sun.m = mSun
earth.m = mEarth
jupiter.m = mJupiter
earth.v=vec(-2000,30500,0) 
sun.v=vec(0,0,0)
jupiter.v=vec(-5000, 10000, 0)
eTrail = curve(color=color.blue)
jTrail = curve(color=color.yellow)
sTrail = curve(color=color.white)

dt=0.005*yR 
t=0.0
print("Nasa Report")
while (t < 20*yR):
    rate(20) 
    earth.pos += earth.v*dt 
    jupiter.pos += jupiter.v*dt
    sun.pos += sun.v*dt
    Fse=g*sun.m*earth.m*(sun.pos-earth.pos)/pow(mag(sun.pos-earth.pos),3) 
    Fsj=g*sun.m*jupiter.m*(sun.pos-jupiter.pos)/pow(mag(sun.pos-jupiter.pos),3)
    Fje=g*jupiter.m*earth.m*(jupiter.pos-earth.pos)/pow(mag(jupiter.pos-earth.pos),3)
    earth.v += ((Fse/mEarth)+(Fje/mEarth))*dt
    sun.v += ((-Fse/mSun)+(-Fsj/mSun))*dt
    jupiter.v += ((Fsj/mJupiter)+(-Fje/mJupiter))*dt
    eTrail.append(earth.pos)
    jTrail.append(jupiter.pos)
    sTrail.append(sun.pos)
    if (t)%(.2*yR)==0:
        print(f"\nYEAR - {t/yR}\nFORCES:\nFse - {mag(Fse):e} N, Fsj - {mag(Fsj):e} N, Fje - {mag(Fje):e} N")
        print(f"VELOCITIES:\nSUN - {mag(sun.v):.1f} KM/H, EARTH - {mag(earth.v):.1f} KM/H, JUPITER - {mag(jupiter.v):.1f} KM/H")
        print(f"DISTANCE FROM SUN:\nEARTH - {(mag(earth.pos-sun.pos)/aU):.2f} AU, JUPITER - {(mag(jupiter.pos-sun.pos)/aU):.2f} AU")
    t+=dt 
print(f"Done - {t}")
```
