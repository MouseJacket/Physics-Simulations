```python
from vpython import *
from math import *
scene = canvas()

shipM = 1000
shipF = 900000 #9800 by default
lY = 9.461e15
c = 299792458
yR = 3.15e7 #Length of the year in seconds 
dist = 4.3 #4.3 by default
gamma = 0.0

#creating ship object
shipHelm = sphere(pos=vec(0,0,0),size=vec(0.1*lY,0.02*lY,0.1*lY),color=color.white)
shipNeck = box(position=vec(0,0,0),size=vec(.015*lY, 0.035*lY, 0.01*lY),color=color.white)
shipNeck.rotate(angle=pi/5, axis=vec(0, 0, -1.5))
shipNeck.pos = vec(-.012*lY,-.0125*lY,0)
shipBody = sphere(pos=vec(-.05*lY,-.025*lY,0),size=vec(0.1*lY,0.025*lY,0.025*lY),color=color.white)
thrusterLeft = cylinder(pos=vec(-0.125*lY,0,-0.018*lY),size=vec(0.07*lY,0.01*lY,0.01*lY),color=color.white)
thrusterRight = cylinder(pos=vec(-0.125*lY,0,0.018*lY),size=vec(0.07*lY,0.01*lY,0.01*lY),color=color.white)
supportRight = box(pos=vec(0,0,0),size=vec(0.005*lY,0.03*lY,0.002*lY),color=color.white)
supportLeft = box(pos=vec(0,0,0),size=vec(0.005*lY,0.03*lY,0.002*lY),color=color.white)
supportRight.rotate(angle=pi/5, axis=vec(-1, 0, 1))
supportLeft.rotate(angle=pi/5, axis=vec(1, 0, 1))
supportRight.pos=vec(-0.075*lY,-0.01*lY,-0.01*lY)
supportLeft.pos=vec(-0.075*lY,-0.01*lY,0.01*lY)
ship = compound([shipHelm,shipNeck,shipBody,thrusterLeft,thrusterRight,supportRight,supportLeft]) #Make it so.
shipSize=ship.size.x

ship2 = pyramid(pos=vec(0,0,0),size=vec(0.1*lY,0.1*lY,0.1*lY),color=color.orange, opacity = 0.3) #Ship 2 is unaffected by the laws of special relativity
earth = sphere(pos=vec(0,0,0),size=vec(0.2*lY,0.2*lY,0.2*lY),texture=textures.earth)
alphaCenturi = sphere(pos=vec(lY*dist,0,0),size=vec(0.4*lY,0.4*lY,0.4*lY),color=color.red,opacity=0.3)

ship.v = 0.0
ship.p = 0.0
ship.m = shipM
ship.f = shipF

ship2.v = 0.0
ship2.p = 0.0
ship2.m = shipM
ship2.f = shipF

scene.range=3*lY
scene.autoscale=0
scene.center = ship.pos

dt=0.05*yR 
t=0.0
tShip=0.0
maxV=0.0
while (ship.pos.x<lY*dist):
    rate(20)
    if (mag(ship.pos))>=((dist*lY)/2):
        ship.f = (abs(ship.f)*-1)
    gamma = 1/sqrt(1-(ship.v/(c))**2)
    ship.p += ship.m*ship.f
    ship.v = (ship.p/ship.m)/sqrt(1+(ship.p/(ship.m*c))**2)
    ship.pos += vec(ship.v*dt,0,0)
    ship.size.x = shipSize/gamma
    scene.center = ship.pos
    if ship.v>maxV:
        maxV = ship.v

    if (ship2.pos.x<=lY*dist*0.98):
        if (mag(ship2.pos))>=((dist*lY)/2):
            ship2.f = (abs(ship2.f)*-1)
        ship2.p += ship2.m*ship2.f
        ship2.v = ship2.p/ship2.m
        ship2.pos += vec(ship2.v*dt,0,0)
    
    t+=dt
    tShip+=dt/gamma

    if (t)%(.5*yR)==0:
        print(f"Ship Pos - {mag(ship.pos)/lY:.2f} LY ({mag(ship.pos)/(lY*dist):.2f}%), V - {ship.v:.2e} ({ship.v/c:.2f} C), F - {ship.f}, P - {ship.p:.2e}, Gamma - {gamma:.3f}")
    
print(f"Completed trip.\nShip time - {tShip/yR:.2f} years, Earth perspective - {t/yR:.2f} years\nMax speed reached - {maxV:.2e} m/s ({maxV/c:.2f} C), Max gamma - {1/sqrt(1-(maxV/(c))**2):.2f}")
```
