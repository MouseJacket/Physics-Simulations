```python

scene = canvas()
scene.autoscale=0

lY = 9.461e15
c = 299792458
aU = 1.5e11 
yR = 3.15e7 #Length of the year in seconds 
g = 6.6743e-11
i = 1
stars = []
dt=0.01 
t=0.0
tmax=30.0
g1=graph(xmin=0,xmax=tmax)

class star: #Decided to create a single star object instead of having to program in the same calculations over and over again.
    def __init__(self, name="unnamed star", size=0.5, r=vec(0,0,0), c=color.white, m=1, v=vec(0,0,0)):
        self.name = name
        self.shape = sphere(pos=r,size=vec(size,size,size),color=c,make_trail=True,trail_radius=0.05,trail_type='curve',interval = 10)
        self.m = m
        self.v = v
        self.pos = r
        self.origin = r
        self.fnet=vec(0,0,0)
        self.k=0
        self.u=0
        self.e=0
        self.kline=gcurve(graph=g1,color=c)
        self.uline=gcurve(graph=g1,color=c)
        self.totline=gcurve(graph=g1,color=c)
        stars.append(self)
    def FNet(self):
        fnet=vec(0,0,0)
        for i in range(len(stars)):
            if stars[i]==self:
                continue
            else:
                body=stars[i]
                distance=self.pos-body.pos
                fnet += (-1*(self.m*body.m)/(mag(distance)**2))*distance
        self.fnet=fnet
        return fnet
    def udPos(self, r):
        self.shape.pos += r
        self.pos += r
    def Eplot(self):
        self.k = 0.5*self.m*(mag(self.v)**2)
        self.u = 0.0
        for i in range(len(stars)):
            if stars[i]!=self:
                body=stars[i]
                self.u+=-1*((self.m*body.m)/mag(self.pos-body.pos))
        self.e=self.u+self.k
        self.kline.plot(pos=(t,self.k))
        self.uline.plot(pos=(t,self.u))
        self.totline.plot(pos=(t,self.e))
        
star1 = star(r=vec(i*1,0,0),c=color.blue,m=1,v=vec(0,0.5,0),name="Blue Star")
star2 = star(r=vec(i*-1,0,0),c=color.red,m=1,v=vec(0,-0.5,0),name="Red Star")
star3 = star(r=vec(i*-10,0,0),c=color.yellow,m=1,v=vec(0.5,0,0),name="Yellow Star")
star4 = star(r=vec(0,i*-10,0),c=color.green,m=2,v=vec(0,0.25,0),name="Greg") #Just for fun

print("Nasa Report")
print(stars)
while (t < tmax):
    rate(100) 
    for i in range(len(stars)):#Variable Update methods. Has to be in seperate loop so that all stars enter next loop with updated variables.
        stars[i].FNet()
    for i in range(len(stars)):
        stars[i].v += (stars[i].fnet*(dt)/stars[i].m)
        stars[i].udPos(stars[i].v*(dt))
        stars[i].Eplot()
    t+=dt 
    if -0.01<=t%(1)<=0.01:
        print(f"t-{t:.1f}")
        for i in range(len(stars)):
            print(f"{stars[i].name} [Fnet - {mag(stars[i].FNet()):.4f} N, dR - {mag(stars[i].pos-stars[i].origin):.4f}, v - {mag(stars[i].v):.4f}, k - {stars[i].k:.4f}, u - {stars[i].u:.4f}, e - {stars[i].e:.4f}")
        print("")
print(f"Done - {t}")
```
