```python
from vpython import *

# 1. Create a graph object (optional, VPython auto-creates one if needed)
#    Specifying one explicitly allows for titles and axis labels
my_graph = graph(title='<b>My Bar Graph</b>', 
                 xtitle='X Value', 
                 ytitle='Y Value',
                 width=600, 
                 height=300)

# 2. Create a gvbars object
#    delta specifies the width of the bars (default is 1)
bars = gvbars(color=color.cyan, delta=0.7)

# 3. Plot data points
#    You can plot data dynamically in loops, common for physical simulations.
#    For a static graph, just add all points:
data_points = [(1, 5), (2, 8), (3, 4), (4, 10)]

for x, y in data_points:
    bars.plot(pos=(x, y))
```
