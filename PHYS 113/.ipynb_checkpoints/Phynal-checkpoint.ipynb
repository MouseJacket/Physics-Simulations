```python
# Import modules here
from vpython import *
from math import *
import random #Will be useful if we want the game to randomly select different arrow types.
import time
import asyncio
scene = canvas()
scene.title = "<b>Phys-Archery</b>"
scene.caption = "<b>W</b> - <i>pitch up</i>, <b>S</b> - <i>pitch down</i>, <b>SPACE</b> - <i>pull back</i>"
scene.autoscale=0
scene.background = vec(0.35, 0.4, .6) 

#Define all variables up here
dt=0.07
t=0.0
g = 9.8
slowC = 7
groundHeight = -5
physObjs = [] #List of all physics objects
arrows = [] #List of all arrows
totalPoints = 0
arrowFlying = False
targetPos = vec(50,0,0)
targetWidth = 3
targetHeight = 3
arrowsRemaining = 5
aimState = True
drawing = False
keys = {}
sensitivity = 4#How fast the bow aims
currentArrow = None
gamePlaying = True
arrowTrail = curve(color = color.red, opacity = 0.5)
attributes = ["AltPlanet","SlowLight", "Antigrav", "Heavy", "Bouncy", "Acceleration"]

#Define classes here
class PhysObj: #Base physics class that other classes inherit variables and methods of
    def __init__(self, grav = True, v = vec(0,0,0), r=vec(0,0,0), name='Unnamed Object',m=1):
        self.name = name
        self.grav = grav #Boolean checking if object is affected by gravity
        self.v = v
        self.v0 = v #Initial velocity
        self.r = r
        self.r0 = r #Initial position
        self.rPre = r #Delayed update R to check for inter-frame hit.
        self.m = m
        self.queuedForce = vec(0,0,0) #Will calculate this force in next relevant frame
        physObjs.append(self)
        
    def FNet(self): #calculates net force on object. Uses queued force variable.
        fnet = vec(0,0,0)
        fnet+=self.queuedForce
        self.queuedForce = vec(0,0,0)
        if self.grav:
            fnet+=vec(0,-g,0)*dt
        return fnet            
        
    def UdPos(self,r): #Updates position
        self.shape.pos += r
        self.r += r
        
    def PosMath(self): #Runs every frame to calculate changes to position
        self.v += ((self.FNet()*dt)/self.m)
        self.UdPos(self.v*dt) 

class BowArrow(PhysObj): #Arrow class, inherits from PhysOb. for different types of arrows, we can easily inherit from this class.
    def __init__(self,v=vec(0,0,0),r=vec(0,1,0),trail=False,name='Normal Arrow',m=1,axis=vec(1,0,0),c=vec(0.7,0.5,0.2),t=textures.wood_old,sharpness=0.3,attribute=None):
        global currentArrow, g
        super().__init__(grav=True,v=v,r=r,name=name,m=m) #Runs inheritted initialization function
        self.shape=arrow(pos=r,axis=axis,color=c,round=True,make_trail=trail,trail_type='points',interval=15,texture=t)
        self.sharpness=sharpness 
        self.direction = vec(0,0,0)
        self.state = 0 #0-hasn't been fired yet, 1-currently flying, 2-has hit the target and embedded, 3- has missed the target or not embedded
        arrows.append(self) #Adds itself to list of arrows
        currentArrow = self
        self.attribute = attribute
        self.mark = None
        scene.title= f"Arrow Attribute - <b>{self.attribute}</b>     Remaining Arrows - <b>{arrowsRemaining}</b>     Total Points - <b>{totalPoints}</b>"
        match attribute:
            case None:
                pass
            case "Antigrav":
                self.name = "Antigrav Arrow"
                self.shape.color = color.purple
            case "Heavy":
                self.m += random.uniform(0.5,4.0)
                self.name = "Heavy Arrow"
                scene.title += f",     M = <b>{self.m*15:.1f} g</b> (<i>{self.m:.1f}X normal weight</i>)"
                self.shape.color = color.gray(1-(self.m/6))
            case "Bouncy":
                self.name = "Bouncy Arrow"
                self.shape.color = color.yellow
            case "Acceleration":
                self.name = "Rocket Arrow"
                self.shape.color = color.green
            case "SlowLight":
                self.name = "Slow Light Arrow"
                self.shape.color = color.blue
                slowC = random.randint(6,15)
                scene.title += f",     C - <b>{slowC} m/s</b>"
                print(f"Speed of light reduced to {slowC} m/s")
            case "AltPlanet":
                self.planet = random.randint(1,8)
                match self.planet:
                    case 1:
                        self.name = "Moon Arrow"
                        scene.background = color.black
                        g = 1.62
                    case 2:
                        self.name = "Mercury Arrow"
                        scene.background = color.gray(0.1)
                        g = 3.7
                    case 3:
                        self.name = "Venus Arrow"
                        scene.background = vec(0.9, 0.5, 0.25)
                        g = 8.87
                    case 4: 
                        self.name = "Martian Arrow"
                        scene.background = vec(0.7, 0.3, 0.2)
                        g = 3.73
                    case 5:
                        self.name = "Jupiter Arrow"
                        scene.background = vec(0.6,0.4,0.1)
                        g = 24.79
                    case 6:
                        self.name = "Saturn Arrow"
                        scene.background = color.yellow
                        g = 10.44
                    case 7:
                        self.name = "Uranus Arrow"
                        g= 8.69
                        scene.background = color.white
                    case 8:
                        self.name = "Neptune Arrow"
                        g = 11.15
                        scene.background = color.blue
                scene.title += f",     Planet - {self.name.split()[0]} (<i>G - <b>{g} m/s/s</b></i>)"
            case _:
                pass
        
    def UdPos(self,r=None): #Updated version of Update Position for arrows
        if self.state != 0:
            super().UdPos(r) #Runs inheritted UdPos() function
            if mag(self.v)>0: #Avoiding division by 0
                self.direction = self.v/mag(self.v)
            self.shape.axis = self.direction
        else:
            self.grav = False
            if mag(bow.axis)>0:
                self.direction = bow.axis/mag(bow.axis)
            self.shape.axis = self.direction
            self.shape.pos = bow.pos
            
    def FNet(self):
        fnet = super().FNet() #Gets fnet from inheritted fnet calculation
        if self.r.y+(self.shape.headlength*self.shape.axis.y) <= groundHeight: #Tests whether arrow has hit ground. May need to work on better collision detection later.
            fnet = ((-1*self.direction)*mag(self.v))/self.sharpness
        return fnet         
        
    def HitDetect(self):
        global totalPoints
        if self.rPre.x<targetPos.x and self.r.x>=targetPos.x:
            if ((targetPos.z-(targetWidth/2)) <= self.r.z <= (targetPos.z+(targetWidth/2))):
                if (targetPos.y - (targetHeight/2)) <= self.r.y <= (targetPos.y + (targetHeight/2)):
                    self.r.x = target.pos.x
                    self.shape.pos.x = target.pos.x
                    points = ceil(3/(abs(((self.r.z-targetPos.z)*targetWidth)-((self.r.y-targetPos.y)*targetHeight))))
                    totalPoints+=points
                    self.state = 2
                    self.shape.color=color.red
                    print(f"Target Hit for {points} points! Distance Traveled - {mag(self.r)-mag(self.r0):.2f}, Final Speed = {mag(self.v):.2f} m/s")
                    self.v *= 0
                    self.grav=False #Just to test
                    Reset()
        elif self.r.y+self.shape.headlength*self.direction.y <= groundHeight: #For ground
            if self.attribute == "Bouncy":
                if self.r.x < targetPos.x:
                    self.v.y *= -0.8
                    if self.v.x<=0.1:
                        print (f"Target Missed - 0 points. Distance Traveled - {mag(self.r)-mag(self.r0):.2f}, Final Speed = {mag(self.v):.2f} m/s")
                        self.state = 3
                        self.mark = label(pos=self.r, yoffset = 20, text = self.name, opacity = 0.1)
                        Reset()
                else:
                    print (f"Target Missed - 0 points. Distance Traveled - {mag(self.r)-mag(self.r0):.2f}, Final Speed = {mag(self.v):.2f} m/s")
                    self.state = 3
                    self.mark = label(pos=self.r, yoffset = 20, text = self.name, opacity = 0.1)
                    Reset()
            else:
                print (f"Target Missed - 0 points. Distance Traveled - {mag(self.r)-mag(self.r0):.2f}, Final Speed = {mag(self.v):.2f} m/s")
                self.mark = label(pos=self.r, yoffset = 20, text = self.name, opacity = 0.1)
                self.state = 3
                Reset()#arrowsRemaining,currentArrow,arrows)
        elif self.attribute == "Antigrav":
            if self.r.x > targetPos.x*1.5:
                print (f"Target Missed - 0 points. Distance Traveled - {mag(self.r)-mag(self.r0):.2f}, Final Speed = {mag(self.v):.2f} m/s")
                self.mark = label(pos=self.r, yoffset = 20, text = self.name, opacity = 0.1)
                self.state = 3
                Reset()
        self.rPre = self.r
                
    def Attribute(self):
        match self.attribute:
            case "Antigrav":
                self.grav = False
            case "Acceleration":
                self.queuedForce = self.direction*5
            case "SlowLight":
                #print(sqrt(1-((mag(self.v)**2)/(slowC**2))))
                if mag(self.v)>=0.999*slowC:
                    self.v = 0.999*slowC*self.direction
                gamma = 1/(sqrt(1-((mag(self.v)**2)/(slowC**2))))
                self.v = (slowC*sqrt(1-(1/(gamma**2))))*self.direction
                self.shape.axis = (1/gamma)*self.direction
                scene.caption+=f",   Gamma - {gamma:.3f}"
            case _:
                pass
        
#Define methods here
def Aim(): 
    global currentArrow
    if 'w' in keys or 'up' in keys:
        bow.rotate(axis=vec(0,0,1),angle = (pi/45)*dt)
    if 's' in keys or 'down' in keys:
        bow.rotate(axis=vec(0,0,1),angle = (-pi/45)*dt)
    currentArrow.UdPos()
        
def Draw():
    bow.pull += 0.05*dt
    if bow.pull >= 1.0:
        bow.pull = 1.0
    bow.size = vec(1+(3*bow.pull),4-(0.5*bow.pull),0.25)
    U = 0.5*bow.k*(3*bow.pull)**2
    scene.caption = f"Bow Pull - <b>{ceil(bow.pull*100)}%</b>,     U - {U:.2f} J"
    
def Fire():
    global drawing, arrowFlying, aimState, currentArrow, arrowsRemaining
    F = bow.k*(3*bow.pull)
    bow.size = vec(1,4,0.25)
    bow.pull = 0.0
    currentArrow.state = 1
    currentArrow.queuedForce = currentArrow.direction*F
    print (f"{currentArrow.name} fired with {F:.2f} N of force and a pitch of {(bow.axis.y*90):.1f} degrees")
    currentArrow.grav = True
    arrowTrail.append(bow.pos)
    arrowFlying = True
    aimState = False
    drawing = False
    arrowsRemaining -= 1
    
def Reset():
    global aimState,arrowFlying,currentArrow,gamePlaying,totalPoints, targetPos, arrowsRemaining, g
    time.sleep(2.5)
    scene.center = bow.pos
    arrowTrail.clear()
    if len(arrows)>0:
        currentArrow.shape.make_trail = False
        if currentArrow.attribute == "AltPlanet":
            scene.background = vec(0.35, 0.4, .6) 
            g = 9.8
        currentArrow = None
    if arrowsRemaining > 0:
        exec(f"Arrow{len(arrows)} = BowArrow(attribute = random.choice(attributes))")
        bow.axis = vec(1,0,0)
        aimState = True
        arrowFlying = False
        scene.caption = "<b>W</b> - <i>pitch up</i>, <b>S</b> - <i>pitch down</i>, <b>SPACE</b> - <i>pull back</i>"
    else:
        scene.title = f"Final Score - {totalPoints}"
        print (f"All arrows used! Final score - {totalPoints} points.")
        gameOverScreen = True
        while gameOverScreen:
            tryagain = input ("Play again? (y/n):\n")
            tryagain.lower()
            if tryagain == "y" or tryagain == "yes":
                for i in range(len(arrows)):
                    arrows[i].mark.visible = False
                    arrows[i].shape.visible = False
                    physObjs.remove(arrows[i])
                    arrows[i] = None
                arrows.clear()
                targetPos = vec(random.randint(30,110),random.randint(0,3),0)
                target.pos = targetPos
                print(f"Target Distance - {targetPos.x} m")
                arrowsRemaining = 5
                totalPoints = 0
                exec(f"Arrow{len(arrows)} = BowArrow(attribute = random.choice(attributes))")
                bow.axis = vec(1,0,0)
                aimState = True
                arrowFlying = False
                gameOverScreen = False
            elif tryagain == "n" or tryagain == "no":
                print ("Good Game")
                gamePlaying = False
                gameOverScreen = False
            else:
                print ("Input not recognized. Please try again")

def Frame():
    for i in range(len(physObjs)):
        physObjs[i].PosMath()
    currentArrow.HitDetect()
    try:
        if currentArrow.state == 1:
            scene.center = currentArrow.r
            scene.caption = f"V - <b>{mag(currentArrow.v):.2f} m/s</b>,   X - <b>{currentArrow.r.x:.1f}</b>"
            arrowTrail.append(currentArrow.r)
            currentArrow.Attribute()
    except AttributeError:
        pass
        
#Write code here
""" Tests
Arrow1 = BowArrow(v=vec(1,1,0.1),r=vec(0,1,0), trail=True)
Arrow2 = BowArrow(v=vec(1,3,-0.3),r=vec(0,1,0), trail=True)
Arrow3 = BowArrow(v=vec(5,1.42,0),r=vec(0,1,0),trail=True) #This one hits
Arrow4 = BowArrow(v=vec(1,-0.1,-0.15),r=vec(0,1,0),trail=True,c=color.purple,name="Anti-Grav Arrow",grav=False) #Test
"""
for i in range(len(attributes)):
    attributes.append(None) #Makes getting an attribute a 50/50 chance
    
archerRange = box(pos=vec(42,groundHeight,0),size = vec(110,0.5,20),texture=textures.rough)
floor = box(pos=vec(42,groundHeight-0.5,0),size = vec(500,0.5,500), color = vec(0.3, 0.45, 0.3))

target=box(pos=targetPos,size = vec(0.2,targetHeight,targetWidth),color = vec(1,0.8,0.8))
bow = box(pos = vec(0,1,0), size = vec(1,4,0.25))
bow.k = 70
bow.pull = 0.0
Reset()
print(f"Target Distance - {targetPos.x} m")
while gamePlaying:
    rate(80)
    #dt = time.time() - t
    t += dt
    keys = keysdown()
    if aimState:
        Aim()
        if ' ' in keys:
            Draw()
            drawing = True
        elif drawing == True:
            Fire()
    else:
        Frame()
        
```
