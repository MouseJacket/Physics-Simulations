```python
from vpython import *
```

```python
scene = canvas()

k=abs(float(input("What is the spring constant (N/m)?\n"))) 
m=abs(float(input("What is the mass of the ball (Kg)?\n")))
x=float(input("What is the initial displacement of the ball (M)?\n"))

L=3  #The unstretched length of the spring 
yceiling=1.5 #The position of the ceiling 
g=9.8 
y0=yceiling-L 
oscFlag = False
oscCount = 0
oscT = 0.0
oscAverageT = 0.0
solidCeiling = False #Adds bounce, not currently working
ceiling=box(pos=vec(0,yceiling,0),length=6,width=4,height=0.2,texture=textures.granite)
scene.autoscale=0
spring1=helix(pos=vec(0,yceiling,0),axis=vec(0,y0+x,0),radius=0.1*L)
ball1=sphere(pos=vec(0,y0+x,0),radius=0.2*L) 
eqlbrm = sphere(pos=vec(0,y0,0),radius=.05*L,color=color.red)
start = sphere(pos=vec(0,y0+x,0),radius=.05*L,color=color.blue)

ball1.pos = vec(0,x,0)
ball1.v = vec(0,0,0)

dt = 0.1
t = 0.0
print("Start")
while (t<20):
    rate(10)
    ball1.F = (-1*(m*9.8))+(k*(y0-ball1.pos.y))
    ball1.v.y += (ball1.F/m)*dt
    if (ball1.pos.y+dt*ball1.v.y>=yceiling and solidCeiling == True):
        ball1.pos += vec(0,-1*(dt*ball1.v.y),0)
    else:
        ball1.pos += vec(0,dt*ball1.v.y,0)
    spring1.axis=vec(0,y0+ball1.pos.y,0)
    if (ball1.v.y>0 and oscFlag==False):
        oscFlag = True
    if (ball1.v.y<0 and oscFlag==True):
        oscCount+=1
        if(oscCount>1):
            oscAverageT = (oscAverageT+oscT)/2
        else:
            oscAverageT=oscT
        print (f"Oscilation {oscCount}, t - {t:.2f}s, oscT - {oscT:.2f}s")
        oscT=0.0
        oscFlag=False

    print(f"{t:.2f}s > F = {ball1.F:.2f} N, Pos = {ball1.pos.y:.2f}")
    t += dt
    oscT += dt
print(f"Done. {oscCount} Oscilations, Average Oscilation Time - {oscAverageT:.2f}s")


```
