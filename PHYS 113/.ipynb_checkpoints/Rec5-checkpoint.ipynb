# Task 1 - Orbiting Stars!

#### I've created a system for you - no need to do any extra coding today. The starting system has two solar-mass stars orbiting one another. You will place a third star by clicking on the system (also 1 solar mass, moving at earth's orbital speed in the +x direction). Find a configuration where all three stars remain in a stable orbit around one another. Take a screenshot of this (or picture with your phone) when you've found a successful configuration and include it in your email for this week. 

If you're interested, here are a few aspects of the coding you might be interested in:

- note the units are in AU, solar masses, and years (how? why? does it work?)
- all three objects are appended to the list called "objects". Notice that the way I calculated the forces in the while loop would have worked with any number of objects! This way, if we wanted to include many stars/planets, we wouldn't have to individually calculate the forces of every pair.
- I used 'for' loops to calculate the forces. We won't ask you to use these yet, but if you want to take a look, the first for loop -  "for i in range(len(objects))" - iterates the variable $i$ in steps of 1 from 0 to one less than the length of the list "objects". The for loop that is nested inside of the first - "for k in range(i+1,len(objects)" - does the same thing, but starting with the $(i+1)$ index. This way, we can calculate pairs of forces without repeating any of them!

```python
from vpython import *
my_canvas1 = canvas(title = "my New Canvas", width=600, height=400) #new canvas

M = 1 #1 solar mass
G = 4*pi**2 #G in solar masses, AU and years (rather than kg, meters and seconds)
v_use = 0.5*6.28 #speed (keeping equal to half the earths orbital speed) in AU/year
t = 0 #starting time
tfinal = 50 #run for 50 years
dt = 0.005 #a 100th of a year for the timestep

my_canvas1.range = 10

star1 = sphere(pos = vec(1.0,0,0), radius = 0.2, color = color.red, v=vec(0,v_use,0), m=M, Fsum = vec(0,0,0), make_trail = True) #star 1
star2 = sphere(pos = vec(-1.0,0,0), radius = 0.2, color = color.blue, v=vec(0,-v_use,0), m=M, Fsum = vec(0,0,0), make_trail = True) #star 2

objects = [] #list of all objects in the system for calculation of forces
objects.append(star1) #appending objects to the list
objects.append(star2)

NewStar = False #boolian flag telling us we haven't yet added the third star

while t<tfinal:
    rate(40)
    if NewStar==False and t>=1.5: #letting simulation run for a bit before adding third star
        print("please click location of 3rd star")
        loc = my_canvas1.waitfor('click')
        NewStar = True #changing variable to keep from running this if statement again
        objects.append(sphere(pos=loc.pos, radius = 0.2, color = color.yellow, v=vec(v_use,0,0), m=M, Fsum = vec(0,0,0), make_trail = True)) #3rd star!
    for i in range(len(objects)):
        for k in range(i+1,len(objects)):
            dist = mag(objects[i].pos-objects[k].pos) #distance between stars
            Force1 = -G * (objects[i].m*objects[k].m)*(objects[i].pos-objects[k].pos)/(pow(dist,3)) #calculating force pairs
            objects[k].Fsum += -Force1 #incrementing the sum of forces on each object
            objects[i].Fsum += Force1 #force pair
        objects[i].v += (objects[i].Fsum/objects[i].m)*dt #adjusting the velocity
        objects[i].pos += objects[i].v*dt #adjusting the position
        objects[i].Fsum = vec(0,0,0) #resetting the force
    t+=dt #incrementing the current time
    
```

# Task 2 - Springs!

#### In our recitation assignment last week, we modeled a 1-dimensional spring-mass system. This is an inherently predictable system. This time, we are going to have two springs attached to each other in series with a mass between them (along with the ability to move in more than one direction). See if you can knock the ball off of the pole to the bottom-left of the screen by placing the two masses and seeing what happens (with clicks). After getting the ball to fall off of it's post, take a screenshot (or picture with your phone) and attach it to this week's email submission.

#### Don't worry if the simulation appears to break a bit or looks a little unrealistic - that may happen depending on how chaotic you make the movement of the springs.

If you're interested, here are a few aspects of the coding you might be interested in:

- I've included a number of cases where the masses in the simulation can collide with the barriers in the canvas. I've also (somewhat crudely) included possible collisions between the masses on the springs and the ball at the top of the pole. How did I try to do this? Does it make physical sense?
- The springs can move and exert forces in two dimensions here (this is close to the bottom of the code). If you look at the force calculation, you should notice that it still looks quite like the classic $F = -k\Delta x$. We use the "mag" function to help find the difference between the spring's length and it's unstretched length, and then use the "norm" function to make a unit vector in the direction of the forces.
- There are user inputs for "clicking" into place the position of masses. Can you pinpoint where in the code this happens? How does it work?
- I use a lot of boolian if statements for collisions (with the ground, the ceiling, the pole, and the other masses). Can you tell what some of them are doing and why?

```python
from vpython import * #re-importing in case you want to run this block on it's own

my_canvas2 = canvas(title = "my New Canvas", width=600, height=400, background=vec(0.3,0.3,0.9)) #new canvas
L=3 # The unstretched length of the spring
yceiling=1.5 # The position of the ceiling
g=9.8 #acceleration of gravity
k = 20 #N/m
M = 3 #kg
t=0 #start time
tmax=30 #max time
dt = 0.005 #timestep

my_canvas2.range=10

ceiling=box(pos=vec(0,yceiling,0),length=100,width=100,height=0.2,texture=textures.wood) #make upper ceiling
ground=box(pos=vec(0,-20*yceiling,0),length=1000,width=1000,height=0.2,color=color.green) #make floor (for aesthetic purposes)
pole = cylinder(pos=vec(-10,-20*yceiling,0), axis=vec(0, 18*yceiling, 0), color=color.red, radius = 0.3) #platform for ball
ball = sphere(pos=pole.pos+pole.axis+vec(0,0.5,0),radius=0.5, color=color.yellow, m=M, v=vec(0,0,0)) #ball to hit off of platform

print('click on position of first mass')
pos1 = my_canvas2.waitfor("click") #choosing location of first mass
if pos1.pos.y > yceiling-0.6: #can't place the masses above the ceiling
    pos1.pos.y = yceiling-0.6
if pos1.pos.x < (pole.pos.x + pole.radius + 0.5): #can't place the masses to the left of the pole
    pos1.pos.x = (pole.pos.x + pole.radius + 0.5)
if pos1.pos.y < ground.pos.y + 0.5: #can't place the masses below the ground
    pos1.pos.y = pos.ground.y + 0.5
spring1 = helix(pos=vec(0,yceiling,0), axis =(pos1.pos-vec(0,yceiling,0)), radius=0.3) #create spring and mass based on click
mass1 = sphere(pos = spring1.pos + spring1.axis, radius=0.5, m=M, v=vec(0,0,0),F=vec(0,0,0))

print('click on position of second mass') #choosing location of second mass
pos2 = my_canvas2.waitfor("click") #choosing location of second mass
if pos2.pos.y > yceiling-0.6: #can't place the masses above the ceiling
    pos2.pos.y = yceiling-0.6
if pos2.pos.x < (pole.pos.x + pole.radius + 0.5): #can't place the masses to the left of the pole
    pos2.pos.x = (pole.pos.x + pole.radius + 0.5)
if pos2.pos.y < ground.pos.y + 0.5: #can't place the masses below the ground
    pos2.pos.y = ground.pos.y + 0.5
spring2 = helix(pos=mass1.pos, axis = (pos2.pos-mass1.pos), radius = 0.3) #create second spring and mass based on click
mass2 = sphere(pos=spring2.pos+spring2.axis,radius=0.5, m=M, v=vec(0,0,0),F=vec(0,0,0))

trail = curve(color=color.yellow) #creating a trail for just the ball when it falls off the pole
while t < tmax:
    rate(1.0/dt)

    #various collisions with ceiling and with pole (can break down relatively easily)
    if mass1.pos.y >= yceiling-0.1-mass1.radius and mass1.v.y > 0: #mass collision with ceiling
        mass1.v.y = -mass1.v.y
    if mass2.pos.y >= yceiling-0.1-mass1.radius and mass2.v.y > 0: #mass collision with ceiling
        mass2.v.y = -mass2.v.y
    if mass1.pos.y <= (-2*yceiling - mass1.radius) and mass1.pos.x <= (pole.pos.x+pole.radius+mass1.radius) and mass1.v.x < 0: #mass collision with pole
        if mass1.pos.x > (pole.pos.x):
            mass1.v.x = -mass1.v.x
    if mass2.pos.y <= (-2*yceiling - mass2.radius) and mass2.pos.x <= (pole.pos.x+pole.radius+mass2.radius) and mass2.v.x < 0: #mass collision with pole
        if mass2.pos.x > (pole.pos.x):
            mass2.v.x = -mass2.v.x
    if mass1.pos.y <= (-2*yceiling - mass1.radius) and mass1.pos.x >= (pole.pos.x-pole.radius-mass1.radius) and mass1.v.x > 0: #mass collision with pole
        if mass1.pos.x < (pole.pos.x):
            mass1.v.x = -mass1.v.x
    if mass2.pos.y <= (-2*yceiling - mass2.radius) and mass2.pos.x >= (pole.pos.x-pole.radius-mass2.radius) and mass2.v.x > 0: #mass collision with pole
        if mass2.pos.x < (pole.pos.x):
            mass2.v.x = -mass2.v.x
    if mass1.pos.y <= ground.pos.y + mass1.radius and mass1.v.y < 0:
        mass1.v.y = -mass1.v.y
    if mass2.pos.y <= ground.pos.y + mass2.radius and mass2.v.y < 0:
        mass2.v.y = -mass2.v.y

    #mass collision with other mass
    if mag(mass1.pos - mass2.pos) < (mass1.radius+mass2.radius):
        vdiff = mass1.v - mass2.v
        ptot = mass1.v * mass1.m + mass2.v * mass2.m
        if dot(norm(vdiff),norm(mass2.pos-mass1.pos)) > 0: #are they moving towards eachother?
            dp1 = norm(mass1.pos-mass2.pos) * mag(ptot) * abs(dot(norm(vdiff),norm(mass2.pos-mass1.pos)))
            dp2 = - dp1
            mass1.v += dp1/mass1.m
            mass2.v += dp2/mass2.m
            
    #mass collision with ball
    if mag(mass1.pos - ball.pos) < (mass1.radius+ball.radius) and dot(norm(mass1.pos-ball.pos),norm(mass1.v)) < 0: 
        alignment = -dot(norm(mass1.pos-ball.pos),norm(mass1.v))
        p1 = mass1.v*mass1.m
        if p1.y > 0: #instantaneous exchange of momentum
            deltap = alignment*p1
            ball.v += deltap/ball.m
            mass1.v += -deltap/ball.m
        else: #instantaneous exchange of momentum (only in x-direction)
            deltap = alignment*vec(p1.x,0,0)
            ball.v += deltap/ball.m
            mass1.v += -deltap/ball.m

    #calculation of forces and movement of ball after it has been hit off the pole
    if mag(ball.v) > 0:  
        ball.F = ball.m*vec(0,-g,0)
        ball.v += (ball.F/ball.m)*dt
        ball.pos += ball.v*dt
        trail.append(ball.pos) #creating the trail of the ball
        if ball.pos.y >= yceiling-0.1-ball.radius and ball.v.y > 0: #collision with ceiling just in case
            if ball.pos.x < 0.5*ceiling.width and ball.pos.x > -0.5*ceiling.width:
                ball.v.y = -ball.v.y
        if ball.pos.y <= ground.pos.y + ball.radius and ball.v.y < 0: #ball collision with the ground
            if ball.pos.x < 0.5*ground.width and ball.pos.x > -0.5*ground.width:
                ball.v.y = -ball.v.y

    #force calculations, velocity and position updates for the two spring-masses
    mass1.F = -g*mass1.m*vec(0,1,0) -k*(mag(spring1.axis)-L)*norm(spring1.axis) + k*(mag(spring2.axis)-L)*norm(spring2.axis)
    mass2.F = -g*mass1.m*vec(0,1,0) - k*(mag(spring2.axis)-L)*norm(spring2.axis)
    mass1.v += (mass1.F/mass1.m)*dt
    mass2.v += (mass2.F/mass2.m)*dt
    mass1.pos += mass1.v*dt
    mass2.pos += mass2.v*dt
    
    #adjusting the springs to keep them attached to the masses
    spring1.axis = mass1.pos - spring1.pos
    spring2.pos = mass1.pos
    spring2.axis = mass2.pos - mass1.pos
    
    #incrementing the time
    t += dt
    
```
