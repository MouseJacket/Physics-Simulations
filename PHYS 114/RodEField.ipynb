```python
from vpython import *
scene = canvas()
charge = 1.602e-19
mass = 1.67262192e-27
R = 0.84e-15
oofpez = 9e9
Qtot = -1e-9
N = 30
L = 1
dY = L/N
r0 = -L/2 + dY/2
slices = []
chargeLayers = 1

scene.background = color.white

rod = cylinder(pos=vec(0,-L/2, 0), axis = vec(0,1,0),length = L, radius = dY/1.9, color=color.white, opacity = 0.5)
for i in range(N):
    a = sphere(pos=vector(0,r0+i*dY,0), radius=dY/2, color=color.red, q=Qtot/N)
    slices.append(a) ## add sphere to list

def Obs(yobs = 0.4, layers = 1, R = 0.3):
    observation = []
    dtheta = pi/6
    for layer in range(layers):
        theta = 0
        while theta < 2*pi:
            a = arrow(pos=vector(((layer+1)*R)*cos(theta), yobs, ((layer+1)*R)*sin(theta)), color=color.orange,axis=vector(0,0,0), opacity = 0.55)
            if a.pos.y<=0:
                a.color=color.blue
            observation.append(a)
            theta = theta + dtheta
    ## calculate E at observation location
    sf = 0.002 ## arrow scale factor
    j = 0
    ## outer loop
    while j < len(observation):
        rate(500)
        earrow = observation[j]
        ## add E of all slices for this obs. loc.
        i = 0
        E_net = vector(0,0,0)
        ## inner loop
        while i < N:
            r = earrow.pos - slices[i].pos
            rhat = r/mag(r)
            if i>(N/2):
                E = (oofpez * slices[i].q * -1 / mag(r)**2) * rhat
            else:
                E = (oofpez * slices[i].q / mag(r)**2) * rhat
            E_net += E
            i += 1
        ## end of inner loop
        earrow.axis = sf*E_net
        j += 1
        ## end of outer loop
for i in range(6):
    Obs(yobs=((L/6)*(i+1))-(L/2),layers=2)
    
```
