```python
from vpython import *
oofpez = 9e9 ## One-Over-Four-Pi-Epsilon-Zero
qe = 1.6e-19
sf = 3e-16 ## scale factor
q_source = qe

scene = canvas()

R = 1e-8 ## radius of circle
dtheta = pi/6
dphi = dtheta
thetaMax = 2*pi
phiMax = pi
arrows = []
layers = 3

#I'm not sure if I'm missing why this wasn't the answer everyone was going with. 
#It isn't recalculating each arrow, but it's accomplishing the same thing.
def flip(evt): 
    global q_source
    q_source *= -1
    for arrow in arrows:
        arrow.axis *= -1
        if q_source > 0:
            arrow.color = color.orange
        else:
            arrow.color = color.blue
    
    
def draw():
    for layer in range(layers):
        source = sphere(pos=vector(0,0,0), radius=1e-9, color=color.red)
        phi = 0
        theta = 0
        while phi < phiMax:
            theta = 0
            while theta < thetaMax:
                rate(500)
                r_obs = (R*(layer+1)) * vector(sin(theta)*cos(phi),sin(theta)*sin(phi),cos(theta))+source.pos
                r = r_obs - source.pos
                rhat = r/mag(r)
                E = (oofpez * q_source / mag(r)**2) * rhat
                arrows.append(arrow(pos=r_obs, color=color.orange, axis=sf*E, shaftwidth = source.radius))
                theta += dtheta
            phi += dphi

b1 = button(pos=scene.caption_anchor , bind=flip , text="flip")

draw()
```
