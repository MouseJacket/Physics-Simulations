```python
#12.61.a

from vpython import *
from math import *
from itertools import *

oscillatorsPerAtom = 3
n = 2*oscillatorsPerAtom
q = 4
numMicrostates = factorial(q+n-1)/(factorial(q)*factorial(n-1))
microstates = []
state = [] #Current microstate, needed different name than microstate
atom1 = []
atom2 = []
atoms = [atom1,atom2]
histogram = graph(title="<b>Histogram of Einstein Model of an Atom</b>",xtitle = "Quanta per Atom",ytitle="Microstates")
bar1 = gvbars(color=color.red, label = "Atom 1", legend = True, delta = 0.75)
bar2 = gvbars(color=color.blue, label = "Atom 2", legend = True,delta = 0.75)
bar3 = gvbars(color=color.yellow, label = "Omega", legend = True, delta=0.5)
barData = []

def calcMicro(q,n):
    for combo in combinations_with_replacement(range(n),q):
        state = [0] * n
        for j in combo:
            state[j] += 1
        microstates.append(state)

calcMicro(q,n)
print (f"Number of Microstates - {len(microstates)}")
for i in range(len(microstates)):
    stateVec1 = []
    stateVec2 = []
    for j in range(len(microstates[i])):
        if j <= 2:
            stateVec1.append(microstates[i][j])
        else:
            stateVec2.append(microstates[i][j])
    atom1.append(stateVec1)
    atom2.append(stateVec2)
atom1 = sorted(atom1, key=lambda x: sum(x), reverse=True)
atom2 = sorted(atom2, key=lambda x: sum(x))

for i in range(q+1):
    barData.append([factorial(oscillatorsPerAtom+i)/(factorial(i+1)*factorial(oscillatorsPerAtom-1)),
                   factorial(oscillatorsPerAtom+(q-i))/(factorial((q-i)+1)*factorial(oscillatorsPerAtom-1))])
    
for i in range(len(barData)):
    bar1.plot(i,barData[i][0])
    bar2.plot(i,barData[i][1])
    bar3.plot(i,barData[i][0]*barData[i][1])

for i in range(len(microstates)):
    print(f"{i+1} >  {atom1[i]} - {atom2[i]}")
```

```python
#12.61.c

from vpython import *
from math import *

scene = canvas()

n = 500
q = 100
numMicrostates = factorial(q+n-1)/(factorial(q)*factorial(n-1))
g1 = graph(title="<b>Histogram of 2 Block System</b>",xtitle = "Quanta per Blcok",ytitle="Microstates")
bars1 = gvbars(graph=g1,color=color.yellow, label = "Omega", legend = True, delta=0.5)
g2 = graph(title="<b>Histogram of 2 Block System</b>",xtitle = "Quanta per Blcok",ytitle="Microstates")
bars2 = gvbars(graph=g2,color=color.yellow, label = "Omega", legend = True, delta=0.5)
g3 = graph(title="<b>Histogram of 2 Block System</b>",xtitle = "Quanta per Blcok",ytitle="Microstates")
bars3 = gvbars(graph=g3,color=color.yellow, label = "Omega", legend = True, delta=0.5)

barData = []
print (f"Number of Microstates - {numMicrostates}")

def graph(bar,ratio):
    for i in range(q+1):
        barData.append([factorial(ceil(n*ratio)+i)/(factorial(i+1)*factorial(ceil(n*ratio)-1)), 
                        factorial(floor(n*(1-ratio))+(q-i))/(factorial((q-i)+1)*factorial(floor(n*(1-ratio)-1)))])
    for i in range(len(barData)):
        bar.plot(i,barData[i][0]*barData[i][1])
    barData.clear()

graph(bars1, 1/3)
graph(bars2, 1/6)
graph(bars3, 7/10)
"""
Histogram is naturally skewed towards the object with the less oscillators, meaning that a fraction defining block A
will be about where the peak of entropy falls (since the fraction is in relation to how many of the oscillators block A.)
"""
```

```python
#12.62
from vpython import *
from math import *

scene = canvas()

nBlock1 = 300
nBlock2 = 200
n = nBlock1+nBlock2
q = 100
numMicrostates = factorial(q+n-1)/(factorial(q)*factorial(n-1))
histogram = graph(title="<b>Entropy Graph</b>",xtitle = "Quanta",ytitle="Natural Log Omega")
line1 = gcurve(color=color.blue,label = "Block 1",legend = True)
line2 = gcurve(color=color.red,label = "Block 2", legend = True)
line3 = gcurve(color=color.yellow,label = "Entropy", legend = True)

barData = []

for i in range(q+1):
    barData.append([factorial(nBlock1+i)/(factorial(i+1)*factorial(nBlock1-1)),
                   factorial(nBlock2+(q-i))/(factorial((q-i)+1)*factorial(nBlock2-1))])
print (f"Number of Microstates - {numMicrostates}")

for i in range(len(barData)):
    line1.plot(i,log(barData[i][0]))
    line2.plot(i,log(barData[i][1]))
    line3.plot(i,log(barData[i][0]*barData[i][1]))

#Highest ln at q = 60, ln(omega) represents entropy
```

```python
#12.63
from vpython import *
from math import *

scene = canvas()

nBlock1 = 300
nBlock2 = 200
hBar = 1.054571817e-34
Boltzmann = 1.380649e-23
Ksi=16*4
Ma=4.48e-26 #Mass of an atom of Aluminum
n = nBlock1+nBlock2
q = 100
dE = hBar*sqrt(Ksi/Ma)
numMicrostates = factorial(q+n-1)/(factorial(q)*factorial(n-1))
histogram = graph(title="<b>Heat Exchange Graph</b>",xtitle = "Quanta",ytitle="Natural Log Omega")
line1 = gcurve(color=color.blue,label = "Block 1",legend = True)
line2 = gcurve(color=color.red,label = "Block 2", legend = True)

barData = []

for i in range(q+1):
    barData.append([factorial(nBlock1+i)/(factorial(i+1)*factorial(nBlock1-1)),
                   factorial(nBlock2+(q-i))/(factorial((q-i)+1)*factorial(nBlock2-1))])
print (f"Number of Microstates - {numMicrostates}")

for i in range(len(barData)):
    if (i==0):
        continue
    else:
        dS1 = Boltzmann * (log(barData[i][0])-log(barData[i-1][0]))
        dS2 = Boltzmann * (log(barData[i-1][1])-log(barData[i][1]))
    T1 = dE/dS1
    T2 = dE/dS2
    line1.plot(i,T1)
    line2.plot(i,T2)
    #The crossing point of the lines represents thermal equilibrium. 
    #Point of highest energy, point of similar temperature.
```
